```python
# Задание 1: Средние буквы слова

word = 'test'
length = len(word)
middle = length // 2

if length % 2 == 0:
    # Чётное количество букв - выводим две средние
    print(word[middle - 1:middle + 1])
else:
    # Нечётное количество букв - выводим одну среднюю
    print(word[middle])

print("\nПроверка:")
print("word = 'test' ->", word[len(word)//2 - 1:len(word)//2 + 1])
word2 = 'testing'
print("word = 'testing' ->", word2[len(word2)//2])


# Задание 2: Dating-сервис

boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print("Внимание, кто-то может остаться без пары!")
else:
    boys_sorted = sorted(boys)
    girls_sorted = sorted(girls)
    
    print("Идеальные пары:")
    for i in range(len(boys_sorted)):
        print(f"{boys_sorted[i]} и {girls_sorted[i]}")

print("\n--- Проверка второго примера ---")
boys2 = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Michael']
girls2 = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys2) != len(girls2):
    print("Внимание, кто-то может остаться без пары!")
else:
    boys_sorted = sorted(boys2)
    girls_sorted = sorted(girls2)
    
    print("Идеальные пары:")
    for i in range(len(boys_sorted)):
        print(f"{boys_sorted[i]} и {girls_sorted[i]}")
```
